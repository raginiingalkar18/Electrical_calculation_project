```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
file = "/content/communication_system_fault_dataset.csv"
data = pd.read_csv(file, sep=',', engine='python')
```

```python
print(data.isnull().sum())
```

```python
if len(data.columns) == 1 and ',' in data.columns[0]:
    new_columns = data.columns[0].split(',')
    data = data[data.columns[0]].str.split(',', expand=True)
    data.columns = new_columns
    for col in new_columns:
        data[col] = pd.to_numeric(data[col], errors='ignore')
x = data.drop('Fault_Status', axis=1)
y = data['Fault_Status']
```

```python
x = pd.get_dummies(x, drop_first=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
         x,
         y,
         test_size=0.2,
random_state=42
)
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
# Train Logistic Regreasion Model
Model = LogisticRegression()
Model.fit(x_train, y_train)
```

```python
#prediction
y_pread = Model.predict(x_test)
```

```python
# Evaluation
print(accuracy_score(y_test, y_pread))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pread))
print("\nClassification Report:")
print(classification_report(y_test, y_pread))
```
